###What is Deep Learning?


Deep learning is a subset of machine learning, which is a field dedicated to the study and development of machines that can learn (sometimes with the goal of eventually attaining general artificial intellegence).

###What is Machine Learning?

Machine learning is a subfield of computer science where in machine learn to perform tasks of which they were not explicitly programmed.

###Supervised Machine Learning

Supervised learning is a method for transforming oen dataset into another. For example, if you had a dataset called Monday Stock Prices that recorded the price of every stock on every Monday for the past 10 years, and a second dataset called Tuesday Stock Prices recorded over the same time periode, a supervised learning algorithm might try to use one to predict the otherd\.

###Unsupervised Machine Learning

Unsupervised learning shares a property in common with supervised learning: it transforms one dataset into another. Bot the dataset that it transforms into is not previously known or understood.

###Parametric vs Nonparametric learning

A parametric model is characterized by having a fixed number of parameters, whereas a nonparametric models's number of parameters is infinite (determined by data).

####Parametric Learning

#####Supervised Parametric Learning

Supervised parametric learning machines with a fixed number of knobs (that's the parametric part), wherein learning occurs by turning the knobs.

#####Unsupervised Parametric Learning

Unsupervised parametric learning uses a very similiar approach, suses knobs to group data. but in this case, it usually has several knobs for each of which maps the input data's affinity to that particular group (with exceptions and nuance—this us a high-level description).

####Nonparametric Learning

Nonparametric learning is a class of algorithm wherein the number of parameters is based on data (instead of predefined). THis lends itself to methods that generally count in one way or another, thus increasing the number of parameters based on the number of items being counted within the data.

#####What is Input Data?

#####What is a Prediction?

A prediction is what the neural network tells you, given the input data, such as "given the temperature, it is 0% likely that people will ear sweatsuits today" or "given a baseball player's battin gaverage, he is 30% likely to hit a home run" or "given yesterday's stock price, today's stock price will be 101.52."

#####Is this Prediction always Right?

No, sometimes a neural network will make mistakes, but it can learn from them. For Example, if it predicts too high, it will adjust its weight to predict lower next time, and vice versa.

#####How does the Network Learn?

Trial and error. FIrst, it tries to make a prediction. Then, it sees wether the prediciton was too high or too low. Finally, it changes the weight (up or down) to predict more accurately the next time it sees the same input.

####What does this Neural Network do?

It muyltiplies the input by a weight. It "scales" the input by a certain amount.

```python
weights = [0.1, 0.2, 0]
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0
    for i in range(len(a)):
        output += (a[i] * b[i])
    return output
def neural_network(input, weights):
  pred = w_sum(input, weights)
  return pred

toes= [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

input = [toes[0], wlrec[0], nfans[0]]
pred = neural_network(input, weights)
print(pred)
```
